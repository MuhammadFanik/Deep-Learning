```python
# The input to this function is the architecture - This will create 9 Total Trainable params (6 weights, 3 biases)
def initialize_parameters(layers_dimension):
    np.random.seed(0)
    parameters = {}     # Empty Dictionary
    L = len(layers_dimension)

    # I am taking the initial values of w = 0.1 and b = 0
    for i in range(1, L):
        parameters['W' + str(i)] = np.ones((layers_dimension[i-1], layers_dimension[i]))*0.1
        parameters['b' + str(i)] = np.zeros((layers_dimension[i], 1))

    return parameters
```

```python
initialize_parameters([2, 2, 1])
```

```python
# This will calculate the output for any given neuron
def linear_forward(A_previous, W, b):
    Z = np.dot(W.T, A_previous) + b

    return Z
```

```python
# Main function that will perform the forward Propagation

# X --> my input data, which must be flipped to shape (2, 4), one column per student
# parameters --> the dictionary from initialize_parameters() with a1, b1, w2, b2
def L_layer_forward(X, parameters):

    A = X
    L = len(parameters) // 2

    for l in range(1, L+1):  # Each layer and passes each layers output to the next layer as input
        A_prev = A
        wl = parameters['W' + str(l)]
        bl = parameters['b' + str(l)]
        print("A"+str(l-1)+": ", A_prev)
        print("W"+str(l)+": ", wl)
        print("b"+str(l)+": ", bl)
        print("--"*20)

        A = linear_forward(A_prev, wl, bl)
        print("A"+str(l)+": ", A)
        print("**"*20)

    return A, A_prev
```
