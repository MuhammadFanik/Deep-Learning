```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Hyperparameters
learning_rate = 0.1
beta1 = 0.9
beta2 = 0.999
epsilon = 1e-8
```

```python
# Initial values
weight = 10.0
m = 0.0
v = 0.0
```

```python
# Example gradients
gradients = [4, 2, 1, 3, 0.5, 2, 1, 0.2]

weights = [weight]
```

```python
print("Step | Gradient | m_hat   | v_hat   | Updated Weight")
print("-" * 62)

for t, gradient in enumerate(gradients, start=1):

    # Step 1: Update first moment
    m = beta1 * m + (1 - beta1) * gradient
    # Step 2: Update second moment
    v = beta2 * v + (1 - beta2) * gradient**2
    # Step 3: Bias correction
    m_hat = m / (1 - beta1**t)
    v_hat = v / (1 - beta2**t)
    # Step 4: Update weight
    weight = weight - learning_rate * m_hat / (np.sqrt(v_hat) + epsilon)

    weights.append(weight)

    print(
        f"{t:4} | {gradient:8.2f} | "
        f"{m_hat:7.4f} | {v_hat:7.4f} | {weight:14.5f}")
```

```python
# Plot weight changes
plt.figure(figsize=(10, 5))
plt.plot(range(len(weights)), weights, marker="o")
plt.xlabel("Update Step")
plt.ylabel("Weight Value")
plt.title("Adam Optimizer: Weight Updates")
plt.grid(True)
plt.tight_layout()
plt.show()
```
