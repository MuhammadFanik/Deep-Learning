```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import lineMarkers
from scipy.ndimage import label
```

```python
# Raw data: noisy values
values = np.array([10, 2, 12, 4, 14, 6, 10, 3, 13, 5])
```

```python
# Smoothing factor
beta = 0.9
```

```python
# Store the smoothed values
ewma =  np.zeros(len(values))

# Initialize with the first value
ewma[0] = values[0]
```

```python
# Calculate EWMA
for t in range(1, len(values)):
    ewma[t] = beta * ewma[t-1] + (1-beta) * values[t]
```

```python
print(f"Before EWMA: {values}")
print(f"After EWMA: {np.round(ewma, 2)}")
```

### Plot the comparison

```python
plt.figure(figsize=(10, 5))

plt.plot(values, marker="o", linestyle="--", label="Raw values")
plt.plot(ewma, marker="o", label="After EWMA")

plt.title("Effect of EWMA smoothing")
plt.xlabel("Time step")
plt.ylabel("Value")
plt.legend()
plt.show()
```
